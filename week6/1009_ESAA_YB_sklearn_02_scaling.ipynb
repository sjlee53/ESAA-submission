# 사이킷런 연습문제 (2) — 스케일링

비행기 승객 만족도 데이터의 **Age** 컬럼을 MinMaxScaler와 StandardScaler로 각각 스케일링한다.

핵심은 하나다. **`fit`은 train에만 하고, test는 `transform`만 한다.**
test로 다시 `fit`하면 기준이 달라져서 같은 나이가 두 데이터에서 다른 값이 된다.

---

### 데이터 불러오기

원본 문제는 `files.upload()`로 파일을 올리는 방식인데, 세션이 끊기면 다시 올려야 한다.
깃허브에 올려 둔 파일을 주소로 바로 읽도록 바꿨다. 용량 때문에 `.gz`로 압축해 뒀고, pandas가 알아서 풀어서 읽는다.

```python
import pandas as pd

train = pd.read_csv('https://raw.githubusercontent.com/sjlee53/ESAA-submission/main/week6/data/train.csv.gz')
test = pd.read_csv('https://raw.githubusercontent.com/sjlee53/ESAA-submission/main/week6/data/test.csv.gz')

print('train:', train.shape, ' test:', test.shape)
```

```python
train.head()
```

```python
test.head()
```

## 1. MinMaxScaler

### Question. train 데이터의 Age 컬럼을 MinMax 스케일링하여 얻은 값을 age_mm_scaling 컬럼에 추가한 후, train셋과 같은 기준으로 test 데이터의 Age를 스케일링하고 그 결과를 age_mm_scaling 컬럼에 추가하라.

```python
# 전체 답 (아래 step1~step6을 한 번에 쓴 것)
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit(train[['Age']])

train['age_mm_scaling'] = scaler.transform(train[['Age']])
test['age_mm_scaling'] = scaler.transform(test[['Age']])
```

#### step1. 'MinMaxScaler'를 사용하기 위한 라이브러리를 import하기

```python
from sklearn.preprocessing import MinMaxScaler
```

#### step2. MinMaxScaler 객체 생성

```python
scaler = MinMaxScaler()
```

#### step3. train 데이터의 Age를 2차원 형태로 변환 후 이를 a1이라는 데이터로 저장

```python
# 대괄호 두 개 → 1차원 Series가 아니라 2차원 DataFrame이 된다 (스케일러는 2차원만 받는다)
a1 = train[['Age']]
print(a1.shape)
```

#### step4. MinMaxScaler 객체를 사용하여 데이터를 학습

```python
scaler.fit(a1)

# 학습된 기준 = train의 최솟값과 최댓값
print('최솟값:', scaler.data_min_, ' 최댓값:', scaler.data_max_)
```

#### step5. 학습된 MinMaxScaler 객체를 사용하여 'Age'열의 데이터를 변환하고, 'age_mm_scaling'컬럼에 추가

```python
train['age_mm_scaling'] = scaler.transform(a1)
train[['Age', 'age_mm_scaling']].head()
```

#### step6. test 데이터에서도 'Age'컬럼을 추출하여 2차원 배열 형태로 변환하고, 학습된 MinMaxScaler 객체를 사용하여 test 데이터의 'Age'열을 변환 후 이를 'age_mm_scaling'컬럼에 추가

```python
a2 = test[['Age']]

# fit을 다시 하지 않는다. train에서 배운 기준 그대로 변환만 한다.
test['age_mm_scaling'] = scaler.transform(a2)
test[['Age', 'age_mm_scaling']].head()
```

## 2. StandardScaler

### Question. train 데이터의 Age 컬럼을 sklearn 모듈을 활용하여 정규화 스케일링한 결과를 age_s_scaling 컬럼에 추가한 후, train셋과 같은 기준으로 test 데이터의 Age를 스케일링하여 age_s_scaling 컬럼에 추가하라.

#### step1. 'StandardScaler'를 사용하기 위한 라이브러리를 import하기

```python
from sklearn.preprocessing import StandardScaler
```

#### step2. StandardScaler 객체 생성

```python
scaler2 = StandardScaler()
```

#### step3. train 데이터의 Age를 2차원 형태로 변환 후 이를 a1라는 데이터로 저장

```python
a1 = train[['Age']]
print(a1.shape)
```

#### step4. StandardScaler 객체를 사용하여 데이터를 학습

```python
scaler2.fit(a1)

# 학습된 기준 = train의 평균과 표준편차
print('평균:', scaler2.mean_, ' 표준편차:', scaler2.scale_)
```

#### step5. 학습된 StandardScaler 객체를 사용하여 'Age'열의 데이터를 변환하고, 'age_s_scaling'컬럼에 추가

```python
train['age_s_scaling'] = scaler2.transform(a1)
train[['Age', 'age_s_scaling']].head()
```

#### step6. test 데이터에서도 'Age'컬럼을 추출하여 2차원 배열 형태로 변환하고, 학습된 StandardScaler 객체를 사용하여 test데이터의 'Age'열을 변환 후 'age_s_scaling'컬럼에 추가

```python
a2 = test[['Age']]
test['age_s_scaling'] = scaler2.transform(a2)
test[['Age', 'age_s_scaling']].head()
```

#### train 데이터에 추가된 age_mm_scaling 칼럼과 age_s_scaling 칼럼 확인하기

```python
train.head()
```

#### test 데이터에 추가된 age_mm_scaling 칼럼과 age_s_scaling 칼럼 확인하기

```python
test.head()
```

### 결과 확인

두 스케일러가 실제로 무엇을 맞춰 놓았는지 숫자로 본다.

```python
print('[MinMax]   train 범위 : %.4f ~ %.4f'
      % (train['age_mm_scaling'].min(), train['age_mm_scaling'].max()))
print('[Standard] train 평균 : %.4f, 표준편차 : %.4f'
      % (train['age_s_scaling'].mean(), train['age_s_scaling'].std(ddof=0)))

print()
print('같은 기준으로 변환됐는지 - 52세 한 명을 두 데이터에서 비교')
print('train의 52세 :', train.loc[train['Age'] == 52, 'age_mm_scaling'].iloc[0])
print('test의 52세  :', test.loc[test['Age'] == 52, 'age_mm_scaling'].iloc[0])
```

### 정리

- **MinMaxScaler** : `(값 - 최솟값) / (최댓값 - 최솟값)` → 0~1 사이로 눌러 넣는다.
  Age가 7~85세이므로 13세는 `(13-7)/(85-7) = 0.0769`가 된다.
- **StandardScaler** : `(값 - 평균) / 표준편차` → 평균 0, 표준편차 1로 맞춘다.
  평균 39.38세, 표준편차 15.11이므로 13세는 `(13-39.38)/15.11 = -1.745`가 된다.

**왜 test에 `fit`을 다시 하면 안 되나**

train과 test가 서로 다른 잣대로 재어지기 때문이다. 같은 52세가 두 데이터에서 다른 숫자가 되면
모델이 train에서 배운 내용을 test에 그대로 적용할 수 없다.
이번 데이터는 train과 test의 Age 범위가 둘 다 7~85라서 숫자가 우연히 같게 나오지만,
범위가 다른 데이터에서는 바로 틀린 결과가 된다.

**자주 하는 실수**

`train['Age']`는 1차원 Series라서 스케일러에 그냥 넣으면 에러가 난다.
`train[['Age']]`처럼 대괄호를 두 번 써서 2차원으로 만들어야 한다.
